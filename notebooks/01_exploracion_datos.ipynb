### Revisión de registros con precio cero y descripción faltante

Después de analizar los registros con `UnitPrice = 0`, se revisó el grupo que presenta simultáneamente `Description` y `CustomerID` ausentes.

Se identificaron **1,454 registros** con estas características. Al considerar también los **1,021 registros** analizados previamente en el Caso 1021, se observa que existen **2,475 registros** con `UnitPrice = 0` y `CustomerID` ausente.

Esta revisión permite determinar si estos registros corresponden a operaciones comerciales o si presentan características que justifican su exclusión del dataset procesado.

```python
# Calculamos las unidades asociadas a los registros con precio cero,
# descripción faltante y cantidad positiva
data[
    (data['UnitPrice'] == 0) &
    (data['Description'].isnull()) &
    (data['Quantity'] > 0)
]['Quantity'].sum()
```

```python
# Calculamos las unidades positivas asociadas a los registros
# con precio cero y descripción faltante, excluyendo códigos especiales
data[
    (data['UnitPrice'] == 0) &
    (data['Description'].isnull()) &
    (data['Quantity'] > 0) &
    (~data['StockCode'].isin(['DOT', 'POST', 'M', 'm', 'S', 'PADS'])) &
    (~data['StockCode'].str.startswith('gift_', na=False))
]['Quantity'].sum()
```

### Impacto en el análisis de unidades

Aunque estos registros no generan ingresos debido a que `UnitPrice = 0`, algunos contienen cantidades positivas. Después de excluir los códigos especiales identificados, se contabilizaron **28,137 unidades** asociadas a registros sin descripción.

Estas unidades podrían afectar análisis basados en la cantidad vendida, como la rotación de productos y la evolución de la demanda. Por este motivo, se considera más consistente excluir del dataset procesado los registros que presentan simultáneamente `UnitPrice = 0` y `CustomerID` ausente.

### Conclusión del proceso de revisión

La exploración permitió identificar un total de **2,475 registros con `UnitPrice = 0` y `CustomerID` ausente**. Este grupo incluye tanto los 1,021 registros analizados inicialmente como los 1,454 registros sin descripción.

Debido a que estos registros no presentan un precio de venta ni un identificador de cliente y pueden introducir cantidades que no corresponden a ventas comerciales convencionales, se decide excluirlos del dataset procesado.

Esta decisión se aplicará de forma general en el script `clean_data.py`, en lugar de tratar los 1,021 registros como un caso aislado.
